```python
#!/usr/bin/env python3
"""
VubaPay Helmet QR Code Generator
Generates QR codes that auto-dial USSD codes when scanned with a phone camera.
"""

import qrcode
from PIL import Image, ImageDraw, ImageFont
import os

# ============================================================
# CONFIGURATION
# ============================================================

# The USSD code that will be dialed when scanned
# Format: *CODE*FUNCTION*SUB*DRIVERCODE#
USSD_CODE = "*182*8*1*112299#"

# Colors (VubaPay brand colors)
PRIMARY_COLOR = (0, 112, 60)      # Rwanda green
SECONDARY_COLOR = (252, 191, 0)   # VubaPay yellow/gold
BACKGROUND_COLOR = (255, 255, 255)
TEXT_COLOR = (0, 0, 0)

# QR Code settings
QR_VERSION = 5           # Higher = more data capacity (1-40)
QR_BOX_SIZE = 10         # Pixel size of each QR module
QR_BORDER = 4            # Border size (modules)

# Output settings
OUTPUT_DIR = "vubapay_qr_codes"

# ============================================================
# QR CODE GENERATION FUNCTIONS
# ============================================================

def generate_basic_qr(data, filename, size=400):
    """
    Generate a basic QR code (no styling)
    
    Args:
        data: String to encode (USSD code)
        filename: Output filename
        size: Final image size in pixels
    """
    qr = qrcode.QRCode(
        version=QR_VERSION,
        error_correction=qrcode.constants.ERROR_CORRECT_H,  # High error correction (30% damage tolerant)
        box_size=QR_BOX_SIZE,
        border=QR_BORDER,
    )
    qr.add_data(data)
    qr.make(fit=True)
    
    img = qr.make_image(fill_color="black", back_color="white")
    img = img.resize((size, size), Image.Resampling.LANCZOS)
    img.save(filename)
    print(f"✅ Generated: {filename}")
    return img


def generate_colored_qr(data, filename, size=500):
    """
    Generate a colored QR code with brand colors
    
    Args:
        data: String to encode
        filename: Output filename
        size: Final image size
    """
    qr = qrcode.QRCode(
        version=QR_VERSION,
        error_correction=qrcode.constants.ERROR_CORRECT_H,
        box_size=QR_BOX_SIZE,
        border=QR_BORDER,
    )
    qr.add_data(data)
    qr.make(fit=True)
    
    # Create colored QR
    img = qr.make_image(fill_color=PRIMARY_COLOR, back_color=BACKGROUND_COLOR)
    img = img.resize((size, size), Image.Resampling.LANCZOS)
    img.save(filename)
    print(f"✅ Generated (colored): {filename}")
    return img


def add_instruction_text(base_img, ussd_code):
    """
    Add instruction text below the QR code
    
    Args:
        base_img: QR code image
        ussd_code: The USSD code to display as text
    
    Returns:
        Image with text footer
    """
    # Create a new image with space for text
    text_height = 120
    new_img = Image.new(
        "RGB", 
        (base_img.width, base_img.height + text_height), 
        BACKGROUND_COLOR
    )
    
    # Paste QR code on top
    new_img.paste(base_img, (0, 0))
    
    # Draw text
    draw = ImageDraw.Draw(new_img)
    
    # Use default font (simpler, works everywhere)
    try:
        # Try to get a larger font if available
        font_medium = ImageFont.truetype("/usr/share/fonts/truetype/dejavu/DejaVuSans.ttf", 16)
        font_small = ImageFont.truetype("/usr/share/fonts/truetype/dejavu/DejaVuSans.ttf", 12)
    except:
        font_medium = ImageFont.load_default()
        font_small = ImageFont.load_default()
    
    # Instruction text
    y_start = base_img.height + 10
    
    draw.text(
        (base_img.width // 2, y_start),
        "📱 SCAN WITH YOUR PHONE CAMERA",
        fill=PRIMARY_COLOR,
        anchor="mt",
        font=font_medium
    )
    
    draw.text(
        (base_img.width // 2, y_start + 25),
        "Your phone will automatically dial:",
        fill=TEXT_COLOR,
        anchor="mt",
        font=font_small
    )
    
    # Draw USSD code in a colored box
    # Measure text size (simplified)
    code_text = ussd_code
    code_width = len(code_text) * 12  # Approximate width
    code_x = (base_img.width - code_width) // 2
    code_y = y_start + 50
    
    # Draw background for USSD code
    draw.rectangle(
        [code_x - 10, code_y - 5, code_x + code_width + 10, code_y + 25],
        fill=SECONDARY_COLOR,
        outline=PRIMARY_COLOR,
        width=2
    )
    
    draw.text(
        (base_img.width // 2, code_y),
        code_text,
        fill=PRIMARY_COLOR,
        anchor="mt",
        font=font_medium
    )
    
    # Footer
    draw.text(
        (base_img.width // 2, base_img.height + text_height - 15),
        "VubaPay | Tora, Scanura, Vuba | No Cash, No Accident",
        fill=(100, 100, 100),
        anchor="mt",
        font=font_small
    )
    
    return new_img


def create_helmet_sticker_format(data, filename, size=500):
    """
    Create QR code formatted for helmet sticker (round corners, high contrast)
    
    Args:
        data: USSD code string
        filename: Output filename
        size: QR size in pixels
    """
    qr = qrcode.QRCode(
        version=QR_VERSION,
        error_correction=qrcode.constants.ERROR_CORRECT_H,
        box_size=QR_BOX_SIZE,
        border=2,  # Smaller border for stickers
    )
    qr.add_data(data)
    qr.make(fit=True)
    
    # Create high contrast black and white QR (scans better outdoors)
    img = qr.make_image(fill_color="black", back_color="white")
    img = img.resize((size, size), Image.Resampling.LANCZOS)
    
    # Create a rounded rectangle mask for sticker effect
    mask = Image.new("L", (size, size), 0)
    mask_draw = ImageDraw.Draw(mask)
    mask_draw.rounded_rectangle(
        [(0, 0), (size, size)], 
        radius=40, 
        fill=255
    )
    
    # Apply rounded corners
    output = Image.new("RGB", (size, size), (255, 255, 255))
    output.paste(img, (0, 0), mask)
    
    # Add a subtle green border
    border_draw = ImageDraw.Draw(output)
    border_draw.rounded_rectangle(
        [(2, 2), (size - 3, size - 3)],
        radius=38,
        outline=PRIMARY_COLOR,
        width=3
    )
    
    output.save(filename)
    print(f"✅ Generated (helmet sticker format): {filename}")
    return output


def generate_all_formats(ussd_code, output_dir):
    """
    Generate all QR code formats for different use cases
    
    Args:
        ussd_code: The USSD code to encode
        output_dir: Directory to save outputs
    """
    # Create output directory
    os.makedirs(output_dir, exist_ok=True)
    
    # Sanitize filename (remove * and #)
    safe_name = ussd_code.replace("*", "_").replace("#", "")
    base_filename = f"{output_dir}/{safe_name}"
    
    print("\n" + "="*60)
    print("🎨 VubaPay QR Code Generator")
    print("="*60)
    print(f"📱 USSD Code: {ussd_code}")
    print(f"📁 Output directory: {output_dir}")
    print("="*60 + "\n")
    
    # 1. Basic QR (print on paper, posters)
    basic_img = generate_basic_qr(ussd_code, f"{base_filename}_basic.png", size=400)
    
    # 2. Colored QR (branded)
    colored_img = generate_colored_qr(ussd_code, f"{base_filename}_colored.png", size=500)
    
    # 3. With instruction text (for customer-facing display)
    text_img = add_instruction_text(basic_img, ussd_code)
    text_img.save(f"{base_filename}_with_instructions.png")
    print(f"✅ Generated (with instructions): {base_filename}_with_instructions.png")
    
    # 4. Helmet sticker format (rounded corners, high contrast)
    sticker_img = create_helmet_sticker_format(ussd_code, f"{base_filename}_helmet_sticker.png", size=500)
    
    # 5. Large printable version (for billboards, posters)
    large_qr = generate_basic_qr(ussd_code, f"{base_filename}_large.png", size=1200)
    
    print("\n" + "="*60)
    print("✅ QR Code Generation Complete!")
    print("="*60)
    print("\n📋 Generated files:")
    print(f"   1. {base_filename}_basic.png - Simple black & white")
    print(f"   2. {base_filename}_colored.png - Branded with VubaPay colors")
    print(f"   3. {base_filename}_with_instructions.png - With user guide")
    print(f"   4. {base_filename}_helmet_sticker.png - Round corners for helmet")
    print(f"   5. {base_filename}_large.png - High resolution for print")
    print("\n🔍 Test Instructions:")
    print("   - Open any of these QR codes on your phone screen")
    print("   - Or print them on paper")
    print("   - Scan with your phone's camera (no app needed)")
    print("   - Your phone will automatically dial: " + ussd_code)
    print("\n📌 For helmet use:")
    print("   - Print 'helmet_sticker' version on waterproof vinyl")
    print("   - Size: 4cm x 4cm")
    print("   - Mount on back of helmet, 2cm from bottom edge")
    print("="*60)


def create_simple_qr(ussd_code, output_filename="vubapay_qr.png"):
    """
    Create a single simple QR code quickly
    
    Args:
        ussd_code: The USSD code to encode
        output_filename: Output filename
    """
    qr = qrcode.QRCode(
        version=1,
        error_correction=qrcode.constants.ERROR_CORRECT_H,
        box_size=10,
        border=4,
    )
    qr.add_data(ussd_code)
    qr.make(fit=True)
    
    img = qr.make_image(fill_color="black", back_color="white")
    img.save(output_filename)
    print(f"✅ Simple QR code saved: {output_filename}")
    print(f"📱 Scan with your phone camera to dial: {ussd_code}")
    return img


# ============================================================
# MAIN EXECUTION
# ============================================================

def main():
    """Main function to generate QR codes"""
    
    # You can change this to any USSD code
    # Example: "*182*8*1*4529#"
    custom_code = "*182*8*1*112299#"
    
    print("\n🚀 VubaPay QR Code Generator")
    print("-" * 40)
    print(f"USSD Code to encode: {custom_code}")
    print("-" * 40)
    
    # Option 1: Generate a single simple QR (fastest)
    print("\n📱 Generating simple QR code...")
    create_simple_qr(custom_code, "vubapay_simple_qr.png")
    
    # Option 2: Generate all formats (more comprehensive)
    print("\n🎨 Generating complete set of QR formats...")
    generate_all_formats(custom_code, OUTPUT_DIR)
    
    print("\n✨ Done! Open the generated PNG files to scan them.")
    print("💡 Tip: Display the QR on your phone screen and scan with another phone's camera.")


# Run the generator
if __name__ == "__main__":
    main()
```

```python
#!/usr/bin/env python3
"""
VubaPay QR Code Generator - DIRECT TO KEYPAD VERSION
When scanned: OPENS PHONE KEYPAD IMMEDIATELY
"""

import qrcode
from PIL import Image, ImageDraw, ImageFont
import os

# ============================================================
# CONFIGURATION - CHANGE THIS
# ============================================================
DRIVER_CODE = "112299"  # Your 4-digit driver code
# ============================================================

# The MAGIC format that opens KEYPAD directly
# %23 is # encoded - this makes phone open dialer, not search
KEYPAD_LINK = f"tel:*182*8*1*{DRIVER_CODE}%23"

print("\n" + "="*60)
print("🎯 VUBA PAY - DIRECT TO KEYPAD QR CODE")
print("="*60)
print(f"📞 Driver Code: {DRIVER_CODE}")
print(f"🔗 QR Encodes: {KEYPAD_LINK}")
print("\n✅ When scanned: Phone OPENS KEYPAD directly")
print("❌ Not search engine | ❌ Not browser")
print("="*60 + "\n")

# Create output directory
os.makedirs("vubapay_qr", exist_ok=True)

# ============================================================
# Generate QR Code (Option 1: Simple black & white)
# ============================================================
qr = qrcode.QRCode(
    version=5,
    error_correction=qrcode.constants.ERROR_CORRECT_H,
    box_size=10,
    border=4,
)
qr.add_data(KEYPAD_LINK)
qr.make(fit=True)

# Create and save QR
img = qr.make_image(fill_color="black", back_color="white")
img = img.resize((500, 500), Image.Resampling.LANCZOS)
img.save("vubapay_qr/vubapay_keypad_qr.png")
print("✅ QR Code 1: vubapay_qr/vubapay_keypad_qr.png")

# ============================================================
# Generate QR Code (Option 2: With instructions for users)
# ============================================================
qr2 = qrcode.QRCode(
    version=5,
    error_correction=qrcode.constants.ERROR_CORRECT_H,
    box_size=10,
    border=4,
)
qr2.add_data(KEYPAD_LINK)
qr2.make(fit=True)

img2 = qr2.make_image(fill_color=(0, 112, 60), back_color="white")  # VubaPay green
img2 = img2.resize((400, 400), Image.Resampling.LANCZOS)

# Add instruction text
final_img = Image.new("RGB", (400, 520), "white")
final_img.paste(img2, (0, 0))
draw = ImageDraw.Draw(final_img)

# Draw instructions
try:
    font = ImageFont.truetype("/usr/share/fonts/truetype/dejavu/DejaVuSans.ttf", 14)
    font_bold = ImageFont.truetype("/usr/share/fonts/truetype/dejavu/DejaVuSans-Bold.ttf", 16)
except:
    font = ImageFont.load_default()
    font_bold = ImageFont.load_default()

draw.text((200, 420), "📱 SCAN WITH CAMERA", fill=(0, 112, 60), anchor="mt", font=font_bold)
draw.text((200, 445), "OPENS YOUR PHONE'S KEYPAD", fill=(100, 100, 100), anchor="mt", font=font)
draw.text((200, 470), f"*182*8*1*{DRIVER_CODE}#", fill=(252, 191, 0), anchor="mt", font=font_bold)
draw.text((200, 495), "Just tap CALL to pay", fill=(100, 100, 100), anchor="mt", font=font)

final_img.save("vubapay_qr/vubapay_with_instructions.png")
print("✅ QR Code 2: vubapay_qr/vubapay_with_instructions.png")

# ============================================================
# Generate Helmet Sticker Version (printable)
# ============================================================
qr3 = qrcode.QRCode(
    version=5,
    error_correction=qrcode.constants.ERROR_CORRECT_H,
    box_size=15,
    border=2,
)
qr3.add_data(KEYPAD_LINK)
qr3.make(fit=True)

img3 = qr3.make_image(fill_color="black", back_color="white")
img3 = img3.resize((800, 800), Image.Resampling.LANCZOS)

# Add rounded corners for sticker
mask = Image.new("L", (800, 800), 0)
mask_draw = ImageDraw.Draw(mask)
mask_draw.rounded_rectangle([(0, 0), (800, 800)], radius=60, fill=255)

sticker = Image.new("RGB", (800, 800), "white")
sticker.paste(img3, (0, 0), mask)

# Add green border
border_draw = ImageDraw.Draw(sticker)
border_draw.rounded_rectangle([(5, 5), (795, 795)], radius=55, outline=(0, 112, 60), width=4)

sticker.save("vubapay_qr/vubapay_helmet_sticker.png")
print("✅ QR Code 3: vubapay_qr/vubapay_helmet_sticker.png")

# ============================================================
# Create simple HTML test page
# ============================================================
html_content = f"""<!DOCTYPE html>
<html>
<head>
    <title>VubaPay - Scan to Open Keypad</title>
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <style>
        body {{
            font-family: Arial, sans-serif;
            text-align: center;
            background: linear-gradient(135deg, #006C3C 0%, #004d2a 100%);
            padding: 20px;
            min-height: 100vh;
        }}
        .card {{
            background: white;
            border-radius: 20px;
            padding: 30px;
            max-width: 500px;
            margin: 50px auto;
            box-shadow: 0 10px 30px rgba(0,0,0,0.3);
        }}
        h1 {{ color: #006C3C; }}
        .qr-img {{ width: 300px; height: 300px; margin: 20px auto; }}
        .ussd {{ 
            background: #FCC400; 
            color: #006C3C; 
            padding: 15px; 
            border-radius: 10px;
            font-family: monospace;
            font-size: 24px;
            font-weight: bold;
            margin: 20px 0;
        }}
        .success {{ background: #d4edda; padding: 15px; border-radius: 10px; margin: 20px 0; }}
        .warning {{ background: #fff3cd; padding: 15px; border-radius: 10px; margin: 20px 0; text-align: left; }}
    </style>
</head>
<body>
    <div class="card">
        <h1>🚀 VubaPay</h1>
        <p><strong>SCAN THIS QR CODE WITH YOUR PHONE CAMERA</strong></p>
        
        <img class="qr-img" src="vubapay_keypad_qr.png" alt="VubaPay QR Code">
        
        <div class="ussd">
            *182*8*1*{DRIVER_CODE}#
        </div>
        
        <div class="success">
            ✅ <strong>What will happen?</strong><br>
            Your phone will open the <strong>KEYPAD/DIALER</strong><br>
            with the USSD code already entered!
        </div>
        
        <div class="warning">
            📱 <strong>How to test:</strong><br>
            1. Open your phone's camera<br>
            2. Point at the QR code above<br>
            3. Tap the popup that appears<br>
            4. Your KEYPAD opens with the code ready!<br>
            5. Press CALL to pay
        </div>
        
        <div class="warning">
            🪖 <strong>For Helmet Sticker:</strong><br>
            File: vubapay_helmet_sticker.png<br>
            Print: 4cm x 4cm on waterproof vinyl<br>
            Mount: Center-back of helmet
        </div>
        
        <p style="color: #666; margin-top: 30px;">
            VubaPay | Tora, Scanura, Vuba | No Cash, No Accident
        </p>
    </div>
</body>
</html>
"""

with open("vubapay_qr/test_page.html", "w", encoding="utf-8") as f:
    f.write(html_content)
print("✅ Test page: vubapay_qr/test_page.html")

print("\n" + "="*60)
print("✨ GENERATION COMPLETE!")
print("="*60)
print("\n📁 Files saved in 'vubapay_qr' folder:")
print("   1. vubapay_keypad_qr.png - Simple QR (use this for testing)")
print("   2. vubapay_with_instructions.png - QR with text guide")
print("   3. vubapay_helmet_sticker.png - For printing on helmets")
print("   4. test_page.html - Open in browser to scan from screen")

print("\n" + "🔬"*30)
print("TEST INSTRUCTIONS:")
print("🔬"*30)
print("1. Open 'vubapay_qr/test_page.html' in your browser")
print("2. Pick up your PHONE")
print("3. Open your CAMERA app (not a QR scanner app)")
print("4. Point at the QR code on your computer screen")
print("5. A popup will show: 'Call *182*8*1*112299#'")
print("6. TAP THE POPUP")
print("7. Your KEYPAD OPENS with the code ready!")
print("8. Press CALL to dial the USSD code")
print("\n🎯 That's it! Direct to keypad, no search engine!")
print("="*60)
```
